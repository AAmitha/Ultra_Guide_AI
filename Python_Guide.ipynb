```python
# Operators - Special symbols that perform specific operations on one or more operands. Types of Operators in Python:

# 1. Arithmetic Operators - +, -, *, /, //, %, **
x=10
y=3
addition = x + y
subtraction = x - y
multiplication = x * y
division = x / y
floor_division = x // y
modulus = x % y
exponentiation = x ** y 

# 2. Comparison Operators - ==, !=, >, <, >=, <=
a=5
b=10
equal = a == b
not_equal = a != b
greater_than = a > b
less_than = a < b
greater_than_or_equal = a >= b
less_than_or_equal = a <= b

# 3. Logical Operators - and, or, not
x=True
y=False
logical_and = x and y
logical_or = x or y
logical_not = not x

# 4. Assignment Operators - =, +=, -=, *=, /=, //=
x=10
y=5
x += y  # x = x + y
x -= y  # x = x - y
x *= y  # x = x * y 
x /= y  # x = x / y 
x //= y  # x = x // y

# 5. Bitwise Operators - &, |, ^, ~, <<, >>
a=5  # 0b0101
b=3  # 0b0011
bitwise_and = a & b  # 0b0001
bitwise_or = a | b  # 0b0111
bitwise_xor = a ^ b  # 0b0110
bitwise_not = ~a  # 0b1010
bitwise_left_shift = a << 2  # 0b10100
bitwise_right_shift = a >> 2  # 0b0001

# 6. Identity Operators - is, is not
x=10
y=10
x is y  # True
x is not y  # False

# 7. Membership Operators - in, not in
s="Hello World"
'h' in s  # True
'h' not in s  # False

# 8. Operator  Precedence - The order in which operators are evaluated in an expression. Parentheses have the highest precedence, followed by exponentiation, then multiplication and division, then addition and subtraction, and finally comparison and logical operators.
result = 3 + 4 * 2 / (1 - 5) ** 2  # result is 3 + 4 * 2 / (1 - 5) ** 2 = 3 + 4 * 2 / (-4) ** 2 = 3 + 4 * 2 / 16 = 3 + 0.5 = 3.5

# 9. Operator Associativity
# Left-to-Right Associativity - +, -, *, /, //, %, &, |, ^, <<, >>, ==, !=, >, <, >=, <=
# Right-to-Left Associativity - **, =, +=, -=, *=, /=, //=, &=, |=, ^=, <<=, >>=
# Parentheses can be used to change the order of evaluation and override operator precedence and associativity.
# Example: (3 + 4) * 2 / (1 - 5) ** 2 = 7 * 2 / (-4) ** 2 = 14 / 16 = 0.875

# 10. Ternary Operator - a if condition else b
x=10
y=5
result = "x is greater than y" if x > y else "x is not greater than y"  # result is "x is greater than y" because x > y is True

# 11. Walrus Operator - :=
# The walrus operator (:=) is used to assign a value to a variable as part of an expression. It allows you to both assign and return a value in the same expression.
# Example:
n = 5
result = (n := n + 1) * 2  # result is 12, n is now 6

# 12. Chained Comparison Operators - a < b < c
a=5
b=10
c=15
result = a < b < c  # result is True
```

```python
# Conditional Statements
# Conditional statements are used to perform different actions based on different conditions. The main conditional statements in Python are if, elif, and else.

# if statement
x = 10
if x > 5:
    print("x is greater than 5")

# if-else statement
x = 5
if x > 5:
    print("x is greater than 5")
else:
    print("x is not greater than 5")

# if-elif-else statement
x = 5 
if x > 10:
    print("x is greater than 10")
elif x > 5: 
    print("x is greater than 5 but not greater than 10")
else:
    print("x is not greater than 10") 

# Nested if statement
x = 10
if x > 5:     
    if x < 15:
        print("x is greater than 5 and less than 15")
    else:
        print("x is greater than or equal to 15")
else:
    print("x is less than or equal to 5") 

# Ternary Operator
x = 10
result = "x is greater than 5" if x > 5 else "x is not greater than 5"
print(result)
```

```python

# Loops in Python - Used to execute a block of code repeatedly until a certain condition is met. The main loops in Python are for and while loops.

# For Loop - Used to iterate over a sequence (like a list, tuple, string, etc.) or other iterable objects. The syntax is: for variable in sequence: block of code
for i in range(5):
    print(i)

# While Loop - Used to execute a block of code as long as a specified condition is true. The syntax is: while condition: block of code
i = 0
while i < 5:
    print(i)
    i += 1

# Break Statement - Used to exit a loop. The syntax is: break
for i in range(5):
    if i == 3:
        break
    print(i)

# Continue Statement - Used to skip the current iteration of a loop. The syntax is: continue
for i in range(5):
    if i == 3:
        continue
    print(i)

# Pass Statement - Used to do nothing. The syntax is: pass
for i in range(5):
    if i == 3:
        pass
    print(i)

# Nested Loops - A loop inside another loop. The inner loop is executed for each iteration of the outer loop.
for i in range(3):
    for j in range(3):
        print(i, j)
```

```python
# Strings
# A string is a sequence of characters enclosed in single quotes (' '), double quotes (" "), or triple quotes (''' ''' or """ """). Strings are immutable, meaning they cannot be changed after they are created. You can perform various operations on strings, such as concatenation, slicing, formatting, etc.
s = "Hello, World!"
print(s)
# String Operations - Python provides a wide range of string operations that allow you to manipulate and work with strings. Some common string operations include:
# 1. Concatenation - Adding two or more strings together to form a single string.
s1 = "Hello"
s2 = "World"
s3 = s1 + " " + s2  # s3 is "Hello World"
# 2. Length - Returns the number of characters in the string.
s = "Hello, World!"
length = len(s)  # length is 13
# 3. Indexing - Accessing individual characters in a string using their index. The index starts from 0 for the first character.
s = "Hello, World!"
print(s[0])  # Output: "H"
# 4. Slicing - Extracting a substring from a string using a range of indices.
s = "Hello, World!"
print(s[2:5])  # Output: "llo"
# 5. Repetition - Repeating a string a specified number of times using the * operator.
s = "Hello"
print(s * 2)  # Output: "HelloHello"
# 6. String Formatting - Inserting values into a string using placeholders and the format() method or f-strings.
s = "Hello, {}!"
name = "Alice"
print(s.format(name))  # Output: "Hello, Alice!"
s = f"Hello, {name}!"
print(s)  # Output: "Hello, Alice!"
# 7. String Methods - Python provides a variety of built-in string methods that allow you to manipulate and work with strings. Some common string methods include:
# 1. str.upper() - Converts all characters in the string to uppercase.
s = "Hello, World!"
print(s.upper())  # Output: "HELLO, WORLD!"
# 2. str.lower() - Converts all characters in the string to lowercase.  
s = "Hello, World!"
print(s.lower())  # Output: "hello, world!"
# 3. str.strip() - Removes leading and trailing whitespace from the string.
s = "   Hello, World!   "
print(s.strip())  # Output: "Hello, World!"
# 4. str.split() - Splits the string into a list of substrings based on a delimiter.
s = "Hello, World!"
print(s.split(","))  # Output: ["Hello", " World!"]   
# 5. str.replace() - Replaces all occurrences of a substring with another substring.
s = "Hello, World!"
print(s.replace("World", "Universe"))  # Output: "Hello, Universe!" 
# 6. str.find() - Returns the index of the first occurrence of a substring in the string. Returns -1 if the substring is not found.
s = "Hello, World!"
print(s.find("World"))  # Output: 7
# 7. str.count() - Returns the number of occurrences of a substring in the string.
s = "Hello, World!"
print(s.count("l"))  # Output: 3  
# 8. str.startswith() - Returns True if the string starts with a specified substring, otherwise returns False.
s = "Hello, World!"
print(s.startswith("Hello"))  # Output: True
# 9. str.endswith() - Returns True if the string ends with a specified substring, otherwise returns False.
s = "Hello, World!"
print(s.endswith("!"))  # Output: True
# 10. str.isalpha() - Returns True if all characters in the string are alphabetic, otherwise returns False.
s = "Hello"
print(s.isalpha())  # Output: True
# 11. str.isdigit() - Returns True if all characters in the string are digits, otherwise returns False.
s = "12345"
print(s.isdigit())  # Output: True
# 12. str.isalnum() - Returns True if all characters in the string are alphanumeric (letters and numbers), otherwise returns False.
s = "Hello123"
print(s.isalnum())  # Output: True  
# 13. str.join() - Joins the elements of an iterable (like a list) into a single string, with a specified separator.
s = "-"
print(s.join(["Hello", "World"]))  # Output: "Hello-World"
# 14. str.format() - Formats the string using placeholders and values.
name = "Alice"
age = 30
print("My name is {} and I am {} years old.".format(name, age))  # Output: "My name is Alice and I am 30 years old."
# 15. f-strings - A more concise way to format strings using expressions inside curly braces {}.
name = "Alice"
age = 30
print(f"My name is {name} and I am {age} years old.")  # Output: "My name is Alice and I am 30 years old."
```

```python
# Data Structures in Python - Lists, Tuples, Dictionaries, Sets

# Lists - An ordered, mutable collection of items enclosed in square brackets [ ]. Lists can contain items of different data types and can be nested (i.e., a list can contain another list). You can perform various operations on lists, such as indexing, slicing, appending, removing, etc.
my_list = [1, "Hello", 3.14, [1, 2, 3]]
# List Methods - Python provides a variety of built-in list methods that allow you to manipulate and work with lists. Some common list methods include:
# 1. list.append() - Adds an item to the end of the list.
my_list = [1, 2, 3]
my_list.append(4)  # my_list is now [1, 2, 3, 4]
# 2. list.insert() - Inserts an item at a specified index in the list.
my_list = [1, 2, 3]
my_list.insert(1, "Hello")  # my_list is now [1, "Hello", 2, 3]
# 3. list.remove() - Removes the first occurrence of an item from the list.
my_list = [1, 2, 3, 4]
my_list.remove(3)  # my_list is now [1, 2, 4]
# 4. list.pop() - Removes and returns the item at a specified index from the list. If no index is specified, it removes and returns the last item in the list.
my_list = [1, 2, 3, 4]
my_list.pop()  # my_list is now [1, 2, 3] and returns 4
my_list.pop(1)  # my_list is now [1, 3] and returns 2
# 5. list.index() - Returns the index of the first occurrence of an item in the list.
my_list = [1, 2, 3, 4]
print(my_list.index(3))  # Output: 2
# 6. list.count() - Returns the number of occurrences of an item in the list.
my_list = [1, 2, 3, 4, 3]
print(my_list.count(3))  # Output: 2
# 7. list.sort() - Sorts the list in ascending order.
my_list = [4, 2, 1, 3]
my_list.sort()  # my_list is now [1, 2, 3, 4]
# 8. list.reverse() - Reverses the order of the items in the list.
my_list = [1, 2, 3]
my_list.reverse()  # my_list is now [3, 2, 1] 
# 9. list.clear() - Removes all items from the list.
my_list = [1, 2, 3]
my_list.clear()  # my_list is now []
# 10. list.copy() - Returns a shallow copy of the list.
my_list = [1, 2, 3]
my_list_copy = my_list.copy()  # my_list_copy is now [1, 2, 3]
# 11. list.extend() - Extends the list by appending all the items from another iterable (like another list).
my_list = [1, 2, 3]
my_list.extend([4, 5])  # my_list is now [1, 2, 3, 4, 5]
# 12. list.count() - Returns the number of occurrences of an item in the list.
my_list = [1, 2, 3, 4, 3]
print(my_list.count(3))  # Output: 2
# list slicing - Extracting a portion of the list using a range of indices.
my_list = [1, 2, 3, 4, 5]
print(my_list[1:4])  # Output: [2, 3, 4]
# list operations - You can perform various operations on lists, such as concatenation, repetition, etc.
my_list1 = [1, 2, 3]
my_list2 = [4, 5, 6]
my_list3 = my_list1 + my_list2  # my_list3 is now [1, 2, 3, 4, 5, 6]
my_list4 = my_list1 * 2  # my_list4 is now [1, 2, 3, 1, 2, 3]
```

```python
# Tuples - An ordered, immutable collection of items enclosed in parentheses ( ). Tuples can contain items of different data types and can be nested (i.e., a tuple can contain another tuple). You can perform various operations on tuples, such as indexing, slicing, etc., but you cannot modify a tuple after it is created.
my_tuple = (1, "Hello", 3.14, (1, 2, 3))
# Tuple Methods - Python provides a few built-in tuple methods that allow you to work with tuples. Some common tuple methods include:
# 1. tuple.count() - Returns the number of occurrences of an item in the tuple.
my_tuple = (1, 2, 3, 4, 3)
print(my_tuple.count(3))  # Output: 2
# 2. tuple.index() - Returns the index of the first occurrence of an item in the tuple.
my_tuple = (1, 2, 3, 4, 3)
print(my_tuple.index(3))  # Output: 3
# Tuple slicing - Extracting a portion of the tuple using a range of indices.
my_tuple = (1, 2, 3, 4, 5)
print(my_tuple[1:4])  # Output: (2, 3, 4)
# Tuple operations - You can perform various operations on tuples, such as concatenation, repetition, etc.
my_tuple1 = (1, 2, 3)
my_tuple2 = (4, 5, 6)
my_tuple3 = my_tuple1 + my_tuple2  # my_tuple3 is now (1, 2, 3, 4, 5, 6)
my_tuple4 = my_tuple1 * 2  # my_tuple4 is now (1, 2, 3, 1, 2, 3)
# Note: Since tuples are immutable, you cannot modify a tuple after it is created. However, you can create a new tuple by concatenating or slicing existing tuples.
# my_tuple = (1, 2, 3)
# my_tuple[0] = 10  # This will raise a TypeError because tuples are immutable
# my_tuple = (10, 2, 3) # This is allowed because we are creating a new tuple, not modifying the existing one.
```

```python
# Dictionaries (dict) - An unordered, mutable collection of key-value pairs enclosed in curly braces { }. Example: {'name': 'Alice', 'age': 30}, {1: 'one', 2: 'two'}, {'a': 1, 'b': 2}
my_dict = {"name": "Alice", "age": 30, "city": "New York"}
# Dictionary Methods - Python provides a variety of built-in dictionary methods that allow you to manipulate and work with dictionaries. Some common dictionary methods include:
# 1. dict.keys() - Returns a view object that displays a list of all the keys in the dictionary.
my_dict = {"name": "Alice", "age": 30, "city": "New York"}
print(my_dict.keys())  # Output: dict_keys(['name', 'age', 'city'])
# 2. dict.values() - Returns a view object that displays a list of all the values in the dictionary.
my_dict = {"name": "Alice", "age": 30, "city": "New York"}
print(my_dict.values())  # Output: dict_values(['Alice', 30, 'New York'])
# 3. dict.items() - Returns a view object that displays a list of all the key-value pairs in the dictionary.
my_dict = {"name": "Alice", "age": 30, "city": "New York"}
print(my_dict.items())  # Output: dict_items([('name', 'Alice'), ('age', 30), ('city', 'New York')])  
# 4. dict.get() - Returns the value for a specified key if the key is in the dictionary, otherwise returns a default value (None if not specified).
my_dict = {"name": "Alice", "age": 30, "city": "New York"}
print(my_dict.get("name"))  # Output: "Alice"
print(my_dict.get("country", "Unknown"))  # Output: "Unknown" 
# 5. dict.update() - Updates the dictionary with the key-value pairs from another dictionary or an iterable of key-value pairs.
my_dict = {"name": "Alice", "age": 30}
my_dict.update({"city": "New York", "country": "USA"})  # my_dict is now {"name": "Alice", "age": 30, "city": "New York", "country": "USA"}
# 6. dict.pop() - Removes and returns the value for a specified key from the dictionary. If the key is not found, it raises a KeyError.
my_dict = {"name": "Alice", "age": 30, "city": "New York"}
print(my_dict.pop("age"))  # Output: 30, my_dict is now {"name": "Alice", "city": "New York"}
# 7. dict.popitem() - Removes and returns an arbitrary key-value pair from the dictionary. If the dictionary is empty, it raises a KeyError.
my_dict = {"name": "Alice", "age": 30, "city": "New York"}
print(my_dict.popitem())  # Output: ('city', 'New York'), my_dict is now {"name": "Alice", "age": 30}
# 8. dict.clear() - Removes all key-value pairs from the dictionary.
my_dict = {"name": "Alice", "age": 30, "city": "New York"}
my_dict.clear()  # my_dict is now {}  
# 9. dict.copy() - Returns a shallow copy of the dictionary.
my_dict = {"name": "Alice", "age": 30, "city": "New York"}
my_dict_copy = my_dict.copy()  # my_dict_copy is now {"name": "Alice", "age": 30, "city": "New York"}
# 10. dict.fromkeys() - Creates a new dictionary with keys from an iterable and values set to a specified value (None if not specified).
keys = ["name", "age", "city"]
new_dict = dict.fromkeys(keys)  # new_dict is now {"name": None, "age": None, "city": None}
new_dict = dict.fromkeys(keys, "Unknown")  # new_dict is now {"name": "Unknown", "age": "Unknown", "city": "Unknown"}
# 11. dict.setdefault() - Returns the value for a specified key if the key is in the dictionary, otherwise inserts the key with a specified value and returns that value.
my_dict = {"name": "Alice", "age": 30}
print(my_dict.setdefault("name", "Unknown"))  # Output: "Alice", my_dict is still {"name": "Alice", "age": 30}
print(my_dict.setdefault("city", "Unknown"))  # Output: "Unknown", my_dict is now {"name": "Alice", "age": 30, "city": "Unknown"}
# 12. dict.items() - Returns a view object that displays a list of all the key-value pairs in the dictionary.
my_dict = {"name": "Alice", "age": 30, "city": "New York"}
print(my_dict.items())  # Output: dict_items([('name', 'Alice'), ('age', 30), ('city', 'New York')])
# dictionary operations - You can perform various operations on dictionaries, such as concatenation, repetition, etc.
dict1 = {"name": "Alice", "age": 30}
dict2 = {"city": "New York", "country": "USA"}
dict3 = {**dict1, **dict2}  # dict3 is now {"name": "Alice", "age": 30, "city": "New York", "country": "USA"}
dict4 = dict1.copy()
dict4.update(dict2)  # dict4 is now {"name": "Alice", "age": 30, "city": "New York", "country": "USA"}
```

```python
# Sets (set) - An unordered, mutable collection of unique items enclosed in curly braces { }. Example: {1, 2, 3}, {'a', 'b', 'c'}, {1, 'a', 3.14}
my_set = {1, 2, 3, 4, 5}
# Set Methods - Python provides a variety of built-in set methods that allow you to manipulate and work with sets. Some common set methods include:
# 1. set.add() - Adds an item to the set.
my_set = {1, 2, 3}
my_set.add(4)  # my_set is now {1, 2, 3, 4}
# 2. set.remove() - Removes an item from the set. If the item is not found, it raises a KeyError.
my_set = {1, 2, 3, 4}
my_set.remove(3)  # my_set is now {1, 2, 4}
# 3. set.discard() - Removes an item from the set if it is present. If the item is not found, it does nothing.
my_set = {1, 2, 3, 4}
my_set.discard(3)  # my_set is still {1, 2, 4}
# 4. set.pop() - Removes and returns an arbitrary item from the set. If the set is empty, it raises a KeyError.
my_set = {1, 2, 3, 4}
print(my_set.pop())  # Output: 1, my_set is now {2, 3, 4}
# 5. set.clear() - Removes all items from the set.
my_set = {1, 2, 3, 4}
my_set.clear()  # my_set is now set()
# 6. set.copy() - Returns a shallow copy of the set.
my_set = {1, 2, 3, 4}
my_set_copy = my_set.copy()  # my_set_copy is now {1, 2, 3, 4}
# 7. set.union() - Returns a new set that is the union of the two sets.
set1 = {1, 2, 3}
set2 = {3, 4, 5}  
union_set = set1.union(set2)  # union_set is now {1, 2, 3, 4, 5}
# 8. set.intersection() - Returns a new set that is the intersection of the two sets.
set1 = {1, 2, 3}
set2 = {3, 4, 5}
intersection_set = set1.intersection(set2)  # intersection_set is now {3}   
# 9. set.difference() - Returns a new set that is the difference between the two sets.
set1 = {1, 2, 3}
set2 = {3, 4, 5}
difference_set = set1.difference(set2)  # difference_set is now {1, 2}
# 10. set.symmetric_difference() - Returns a new set that is the symmetric difference between the two sets (i.e., items that are in either set1 or set2 but not in both).
set1 = {1, 2, 3}
set2 = {3, 4, 5}
symmetric_difference_set = set1.symmetric_difference(set2)  # symmetric_difference_set is now {1, 2, 4, 5}
# 11. set.issubset() - Returns True if the set is a subset of another set, otherwise returns False.
set1 = {1, 2}
set2 = {1, 2, 3, 4}
print(set1.issubset(set2))  # Output: True
# 12. set.issuperset() - Returns True if the set is a superset of another set, otherwise returns False.
set1 = {1, 2, 3, 4}
set2 = {1, 2}
print(set1.issuperset(set2))  # Output: True
# 13. set.isdisjoint() - Returns True if the set has no items in common with another set, otherwise returns False.
set1 = {1, 2, 3}
set2 = {4, 5, 6}
print(set1.isdisjoint(set2))  # Output: True
# set operations - You can perform various operations on sets, such as union, intersection, difference, etc.
set1 = {1, 2, 3}
set2 = {3, 4, 5}
union_set = set1.union(set2)  # union_set is now {1, 2, 3, 4, 5}
intersection_set = set1.intersection(set2)  # intersection_set is now {3}
difference_set = set1.difference(set2)  # difference_set is now {1, 2}
symmetric_difference_set = set1.symmetric_difference(set2)  # symmetric_difference_set is now {1, 2, 4, 5}
# Note: Since sets are unordered, the items in a set do not have a specific order, and you cannot access items in a set using an index. However, you can iterate over the items in a set using a for loop or other iterable methods.
my_set = {1, 2, 3, 4, 5}
for item in my_set:
    print(item)
```

```python
# Functions with Arguments - *args / **kwargs
# Functions - A block of reusable code that performs a specific task. Functions can take arguments (inputs) and return values (outputs). You can define your own functions using the def keyword, or use built-in functions provided by Python.
def greet(name):
    return f"Hello, {name}!"
print(greet("Alice"))  # Output: "Hello, Alice!"
# *args - Used to pass a variable number of non-keyword arguments to a function. The arguments are passed as a tuple.
def sum_numbers(*args):
    return sum(args)
print(sum_numbers(1, 2, 3))  # Output: 6
# **kwargs - Used to pass a variable number of keyword arguments to a function. The arguments are passed as a dictionary.
def print_kwargs(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
print_kwargs(name="Alice", age=30)    # Output: name: Alice, age: 30
# You can also combine *args and **kwargs in a function definition to allow for both variable numbers of non-keyword and keyword arguments.
def combined_args(*args, **kwargs):
    print("Non-keyword arguments:", args)
    print("Keyword arguments:", kwargs)
combined_args(1, 2, 3, name="Alice", age=30)  
# Output: Non-keyword arguments: (1, 2, 3), Keyword arguments: {'name': 'Alice', 'age': 30}
# Note: When defining a function with *args and **kwargs, *args must come before **kwargs in the function definition.

# Arguments / Parameters - The values that are passed to a function when it is called. Parameters are the variables that are defined in the function definition, while arguments are the actual values that are passed to the function when it is called.
def greet(name):  # name is a parameter
    return f"Hello, {name}!"
print(greet("Alice"))  # Alice is an argument passed to the greet function

# Default Arguments - You can provide default values for parameters in a function definition. If an argument is not provided for a parameter with a default value, the default value will be used.
def greet(name="Guest"):
    return f"Hello, {name}!"
print(greet())  # Output: "Hello, Guest!"
print(greet("Alice"))  # Output: "Hello, Alice!"  
# Keyword Arguments - You can pass arguments to a function using the parameter names as keywords. This allows you to specify the arguments in any order and makes the code more readable.
def greet(name, age):
    return f"Hello, {name}! You are {age} years old."
print(greet(name="Alice", age=30))  # Output: "Hello, Alice! You are 30 years old." 
# Variable Scope - The region of the program where a variable is defined and can be accessed. Variables defined inside a function are local to that function and cannot be accessed outside of it, while variables defined outside of any function are global and can be accessed from anywhere in the program.
def my_function():
    local_variable = "I am a local variable"
    print(local_variable)  # Output: "I am a local variable"
my_function() 
# print(local_variable)  # This will raise a NameError because local_variable is not defined outside of the function
global_variable = "I am a global variable"
def my_function():
    print(global_variable)  # Output: "I am a global variable"
my_function() 
# print(global_variable)  # Output: "I am a global variable"
# Note: When defining a function, you can specify the scope of the variables using the global keyword. This will make the variables global and accessible from anywhere in the program.
```

```python
# Comprehensions
# Comprehensions - A concise way to create lists, dictionaries, or sets by iterating over an iterable and applying a condition or transformation to each item. The main types of comprehensions in Python are list comprehensions, dictionary comprehensions, and set comprehensions.
# List Comprehensions - A way to create a new list by applying an expression to each item in an iterable, optionally filtering items using a condition. The syntax is: [expression for item in iterable if condition]
squared_numbers = [x**2 for x in range(5)]  # squared_numbers is now [0, 1, 4, 9, 16]
even_numbers = [x for x in range(10) if x % 2 == 0]  # even_numbers is now [0, 2, 4, 6, 8]
# Dictionary Comprehensions - A way to create a new dictionary by applying an expression to each item in an iterable, optionally filtering items using a condition. The syntax is: {key_expression: value_expression for item in iterable if condition}
squared_dict = {x: x**2 for x in range(5)}  # squared_dict is now {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}
even_dict = {x: x for x in range(10) if x % 2 == 0}  # even_dict is now {0: 0, 2: 2, 4: 4, 6: 6, 8: 8}
# Set Comprehensions - A way to create a new set by applying an expression to each item in an iterable, optionally filtering items using a condition. The syntax is: {expression for item in iterable if condition}
squared_set = {x**2 for x in range(5)}  # squared_set is now {0, 1, 4, 9, 16}
even_set = {x for x in range(10) if x % 2 == 0}  # even_set is now {0, 2, 4, 6, 8}
# Note: Comprehensions provide a more concise and readable way to create new lists, dictionaries, or sets compared to using traditional loops and conditional statements. They can also be more efficient in terms of performance, as they are optimized for creating new collections in a single pass through the iterable.
```

```python
# Object-Oriented Programming (OOP) - A programming paradigm that organizes code into objects, which are instances of classes. OOP allows for encapsulation, inheritance, and polymorphism, making it easier to design and maintain complex software systems.
# Class - A blueprint for creating objects. It defines the attributes (data) and methods (functions) that the objects created from the class will have.
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def greet(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."

person1 = Person("Alice", 30)
print(person1.greet())  # Output: "Hello, my name is Alice and I am 30 years old."

# Classes and Objects - A class is a blueprint for creating objects, while an object is an instance of a class. You can create multiple objects from the same class, and each object can have its own unique attributes and behaviors.
class Car:
    def __init__(self, make, model):
        self.make = make
        self.model = model
    def get_info(self):
        return f"{self.make} {self.model}"

car1 = Car("Toyota", "Camry")
print(car1.get_info())  # Output: "Toyota Camry"
car2 = Car("Honda", "Civic")
print(car2.get_info())  # Output: "Honda Civic"
# Note: Classes and objects are often used together to define and implement complex software systems.
# Inheritance - A mechanism in OOP that allows a new class (called a child or subclass) to inherit attributes and methods from an existing class (called a parent or superclass). This promotes code reusability and allows for the creation of more specific classes based on general ones.
class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        return "I am an animal."
class Dog(Animal):
    def speak(self):
        return "Woof!"
class Cat(Animal):
    def speak(self):
        return "Meow!"
dog1 = Dog("Buddy")
print(dog1.speak())  # Output: "Woof!"
cat1 = Cat("Whiskers")
print(cat1.speak())  # Output: "Meow!"
# Polymorphism - A concept in OOP that allows objects of different classes to be treated as objects of a common superclass. It enables a single interface to represent different types of objects, allowing for flexibility and extensibility in code design.
class Shape:
    def area(self):
        pass
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    def area(self):
        return self.width * self.height
class Circle(Shape):    
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius ** 2
rectangle1 = Rectangle(5, 10)
print(rectangle1.area())  # Output: 50
circle1 = Circle(5)
print(circle1.area())  # Output: 78.5
# Note: Polymorphism allows you to write code that can work with objects of different classes through a common interface, making it easier to extend and maintain your code.
# Encapsulation - A principle in OOP that restricts access to certain components of an object and prevents the accidental modification of data. It is achieved by using access modifiers (like private, protected, public) to control the visibility of attributes and methods within a class.
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private attribute
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
    def get_balance(self):
        return self.__balance  # Public method to get the balance
account1 = BankAccount(1000)
account1.deposit(500)
print(account1.get_balance())  # Output: 1500
account1.withdraw(200)
print(account1.get_balance())  # Output: 1300
# Note: Encapsulation helps to protect the internal state of an object and ensures that it can only be modified through well-defined interfaces (methods), promoting data integrity and security in your code.  
# Note: The concepts of OOP (classes, objects, inheritance, polymorphism, encapsulation) are fundamental to designing and implementing complex software systems in Python and other programming languages. They allow for modular, reusable, and maintainable code.
# Inheritance, Polymorphism, and Encapsulation are key principles of OOP that help you design and implement complex software systems efficiently and maintainably.
# Note: OOP allows you to model real-world entities and their interactions in a way that is intuitive and easy to understand, making it a powerful tool for software development.
# Note: OOP is widely used in Python and many other programming languages to create robust and scalable applications, from simple scripts to complex software systems.
#Super() - A built-in function in Python that allows you to call a method from the parent class in a child class. It is commonly used in the context of inheritance to access and extend the functionality of the parent class.
class Parent:
    def __init__(self, name):
        self.name = name
    def greet(self):
        return f"Hello, I am {self.name}."
class Child(Parent):  
    def __init__(self, name, age):
        super().__init__(name)  # Call the __init__ method of the Parent class to initialize the name attribute
        self.age = age
    def greet(self):
        parent_greeting = super().greet()  # Call the greet method of the Parent class
        return f"{parent_greeting} I am {self.age} years old."  # Add the age attribute to the greeting
child1 = Child("Alice", 30)
print(child1.greet())  # Output: "Hello, I am Alice. I am 30 years old."
# Note: The super() function is a powerful tool in OOP that allows you to leverage the functionality of the parent class while still being able to customize and extend it in the child class. It promotes code reuse and helps to maintain a clear and organized class hierarchy.
# Note: When using super(), it is important to ensure that the parent class is properly initialized and that the method you are calling exists in the parent class to avoid errors.
# Note: The super() function can also be used in multiple inheritance scenarios to call methods from multiple parent classes, allowing for more complex class hierarchies and interactions.
```
